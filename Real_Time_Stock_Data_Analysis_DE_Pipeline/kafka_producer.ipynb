```python
import pandas as pd
from kafka import KafkaProducer
from time import sleep
from json import dumps
import json
```

```python
producer = KafkaProducer(bootstrap_servers=['51.21.152.125:9092'], value_serializer=lambda x: dumps(x).encode('utf-8'))
```

```python

producer.send('stock_test', value={'Apache':'Kakfa streaming service'})
```

```python
df = pd.read_csv("data/indexProcessed.csv")
```

```python
while True:
    dict_stock = df.sample(1).to_dict(orient="records")[0]
    producer.send('stock_test', value=dict_stock)
    sleep(1)
```
